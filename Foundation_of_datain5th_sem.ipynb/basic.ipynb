## Data Audit and Data Dictionary

This section audits the student dataset and generates:
- field names and inferred data types
- allowed values (or numeric ranges)
- missingness percentage
- potential primary key candidates
- a metadata summary for the full dataset

```python
import json
from io import StringIO
from pathlib import Path

import pandas as pd

# Read raw dataset lines from the 3rd cell in this notebook.
nb_path = Path("basic.ipynb")
nb_obj = json.loads(nb_path.read_text(encoding="utf-8"))
raw_lines = [line.strip() for line in nb_obj["cells"][2]["source"] if str(line).strip()]

# Build dataframe from tab-separated text.
raw_tsv = "\n".join(raw_lines)
df = pd.read_csv(StringIO(raw_tsv), sep="\t", dtype=str)

# Normalize blank strings as missing.
df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)
df = df.replace("", pd.NA)

# Numeric conversion map for audit calculations.
numeric_targets = ["Maths", "Science", "English", "History", "Total Marks", "Average Score"]
percent_targets = ["Percentage", "Attendance"]

for col in percent_targets:
    if col in df.columns:
        df[col] = df[col].str.replace("%", "", regex=False)

for col in numeric_targets + percent_targets:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")


def infer_type(series: pd.Series) -> str:
    non_null = series.dropna()
    if non_null.empty:
        return "unknown"

    if pd.api.types.is_numeric_dtype(series):
        as_int = (non_null % 1 == 0).all()
        return "integer" if as_int else "float"

    parsed_dates = pd.to_datetime(non_null, errors="coerce", format="mixed")
    if parsed_dates.notna().all():
        return "datetime"

    return "string"


profile_rows = []
for col in df.columns:
    s = df[col]
    missing_count = int(s.isna().sum())
    missing_pct = round((missing_count / len(df)) * 100, 2)
    distinct_count = int(s.nunique(dropna=True))
    inferred = infer_type(s)

    non_null = s.dropna()
    if inferred in {"integer", "float"} and not non_null.empty:
        allowed = f"{non_null.min()} to {non_null.max()}"
    else:
        uniq = sorted(non_null.astype(str).unique().tolist())
        allowed = ", ".join(uniq[:12]) + (" ..." if len(uniq) > 12 else "")

    is_pk_candidate = (missing_count == 0) and (distinct_count == len(df))

    profile_rows.append(
        {
            "Field": col,
            "Inferred Data Type": inferred,
            "Missing Count": missing_count,
            "Missing %": missing_pct,
            "Distinct Values": distinct_count,
            "Allowed Values / Range": allowed,
            "Potential Primary Key": "Yes" if is_pk_candidate else "No",
        }
    )

data_dictionary = pd.DataFrame(profile_rows)

# Dataset-level metadata summary.
pk_candidates = data_dictionary.loc[
    data_dictionary["Potential Primary Key"] == "Yes", "Field"
].tolist()

metadata_summary = pd.DataFrame(
    {
        "Metric": [
            "Row count",
            "Column count",
            "Duplicate rows",
            "Average missingness across fields (%)",
            "Potential primary keys",
            "Categorical fields",
            "Numeric fields",
        ],
        "Value": [
            len(df),
            df.shape[1],
            int(df.duplicated().sum()),
            round(data_dictionary["Missing %"].mean(), 2),
            ", ".join(pk_candidates) if pk_candidates else "None",
            ", ".join(data_dictionary.loc[data_dictionary["Inferred Data Type"] == "string", "Field"]),
            ", ".join(data_dictionary.loc[data_dictionary["Inferred Data Type"].isin(["integer", "float"]), "Field"]),
        ],
    }
)

# Save outputs for reuse.
data_dictionary.to_csv("student_data_dictionary.csv", index=False)
metadata_summary.to_csv("student_metadata_summary.csv", index=False)

print("Audit completed.")
print("Generated files: student_data_dictionary.csv, student_metadata_summary.csv")

print("\n=== Metadata Summary ===")
display(metadata_summary)

print("\n=== Data Dictionary ===")
display(data_dictionary)

print("\n=== Sample Rows (first 5) ===")
display(df.head())
```
