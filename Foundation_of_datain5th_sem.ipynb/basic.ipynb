```python
# Python Basics - Variables, Data Types, Operators, Conditions, Loops, Functions, Collections, Input/Output

# 1. Variables and data types
name = "Alice"
age = 25
height = 5.9
is_student = True
print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Is student:", is_student)
print("Type of name:", type(name))
print("Type of age:", type(age))
print("Type of height:", type(height))
print("Type of is_student:", type(is_student))

# 2. Basic arithmetic and comparison operators
a, b = 10, 3
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulo:", a % b)
print("Exponent:", a ** b)
print("Equal:", a == b)
print("Greater:", a > b)

# 3. Strings
message = "Hello, Python!"
print(message.lower())
print(message.upper())
print(message.replace("Python", "World"))
print(message.split(","))
print("First character:", message[0])
print("Last character:", message[-1])

# 4. Lists
numbers = [1, 2, 3, 4]
numbers.append(5)
numbers.remove(2)
print("List:", numbers)
print("First element:", numbers[0])
print("Length:", len(numbers))

# 5. Tuples
coordinates = (10, 20)
print("Tuple:", coordinates)

# 6. Sets
colors = {"red", "green", "blue", "red"}
print("Set:", colors)

# 7. Dictionaries
student = {"name": "Alice", "age": 25, "course": "Python"}
print("Dictionary:", student)
print("Name from dict:", student["name"])

# 8. Conditional statements
if age >= 18:
    print("Adult")
else:
    print("Minor")

# 9. Loops
for i in range(5):
    print("For loop:", i)

count = 0
while count < 3:
    print("While loop:", count)
    count += 1

# 10. Functions
def add(x, y):
    return x + y

print("Function result:", add(5, 7))

# 11. List comprehension
squares = [x * x for x in range(1, 6)]
print("Squares:", squares)

# 12. Exception handling
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Error:", e)

# 13. Input
user_name = input("Enter your name: ")      
print("Hello,", user_name)
```
