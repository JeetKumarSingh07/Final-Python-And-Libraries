```python
import numpy as np
x=np.array([2,4,6,8,10])
y=np.array([1,3,5,7,9])
expected_value_x=np.sum(x)
expected_value_y=np.sum(y)
print("Expected value of x:", expected_value_x)

mean_x=np.mean(x)
mean_y=np.mean(y)
print("Mean of x:", mean_x)
print("Mean of y:", mean_y)
```

```python
# standard deviation
std_x=np.std(x)
std_y=np.std(y)
print("Standard deviation of x:", std_x)
print("Standard deviation of y:", std_y)
```

```python
# Variance
var_x=np.var(x)
var_y=np.var(y)
print("Variance of x:", var_x)
print("Variance of y:", var_y)
```

```python
# Covariance Between x and y
covariance=np.cov(x,y)[0][1]
print("Covariance between x and y:", covariance)
```

```python
# f Covariance Matrix 
cov_matrix=np.cov(x,y)
print("Covariance matrix:\n", cov_matrix)
```

```python
import numpy as np
import pandas as pd
from IPython.display import display

np.random.seed(42)

dates = pd.date_range(end=pd.Timestamp.today().normalize(), periods=30, freq='D')
advertising_expenditure = np.random.randint(1200, 5200, size=30)
orders = np.round(35 + (advertising_expenditure * 0.018) + np.random.normal(0, 4, size=30)).astype(int)
orders = np.clip(orders, 20, None)

data = pd.DataFrame({
    'Date': dates,
    'Advertising Expenditure': advertising_expenditure,
    'Daily Orders': orders
})

display(data)
```

```python
expected_advertising_expenditure = data['Advertising Expenditure'].mean()
expected_orders = data['Daily Orders'].mean()

print(f'Expected value of Advertising Expenditure: {expected_advertising_expenditure:.2f}')
print(f'Expected value of Daily Orders: {expected_orders:.2f}')
```

```python
print(f"mean of Advertising Expenditure: {data['Advertising Expenditure'].mean():.2f}")
```

```python
print(f'variance of Advertising Expenditure: {data["Advertising Expenditure"].var():.2f}')
```

```python
print(f'standard deviation of Advertising Expenditure: {data["Advertising Expenditure"].std():.2f}')
```
